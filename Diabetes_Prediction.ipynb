```python
df['EmployeeName'] = df['EmployeeName'].str.capitalize()
df[['EmployeeName']].head(5)
```

```python
# Creating a new column age_group

conditions = [
    df['age'] <= 18,
    df['age'] <= 30,
    df['age'] <= 45,
    df['age'] <= 60
]

choices = [
    'Below 18',
    '19-30',
    '31-45',
    '46-60'
]

df['Age_Group'] = np.select(
    conditions,
    choices,
    default='Above 61'
)

df[['age', 'Age_Group']].head(10)
```

```python
# Converting float value of age into int value.
df['age'] = df['age'].astype(int)
df[['age']].head(7)
```
